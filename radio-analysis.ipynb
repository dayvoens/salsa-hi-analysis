```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    "data/SALSA-torre-2026.csv",
    comment="#"
)
```

```python
df['vlsr_kms'] = df['vlsr_mps'] / 1000

df.head()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df['vlsr_kms'], df['amplitude'])

plt.xlabel('VLSR velocity (km/s)')
plt.ylabel('Amplitude')
plt.title('SALSA HI 21 cm spectrum - l=30.0, b=0.0')
plt.grid(True)

plt.savefig("figures/hi_spectrum_l30_b0.png", dpi=200, bbox_inches="tight")

plt.show()
```
